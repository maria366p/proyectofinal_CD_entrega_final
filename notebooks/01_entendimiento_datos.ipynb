```python
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as mtick
import seaborn as sns
from scipy import stats
import statsmodels.formula.api as smf
from Levenshtein import ratio
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore", category=FutureWarning)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.width", 200)
pd.set_option("display.float_format", "{:,.2f}".format)
sns.set_theme(style="whitegrid", context="notebook")
plt.rcParams["figure.dpi"] = 100

ALPHA = 0.05
RANDOM_STATE = 42
RAW_PATH = "../data/raw/Spend_FA_x_Transit_x_User_ID.xlsx"
PROCESSED_PATH = "../data/processed/facturas_anonimizadas.parquet"
MONTO = "Invoice total with tax (CAD)"
MONTO_ORIG = "Invoice Total with Tax Amount"

MESES = ["ene", "feb", "mar", "abr", "may", "jun", "jul", "ago", "sep", "oct", "nov", "dic"]
def fmt_p(p):
    """Valor p en texto, sin notación científica."""
    return "p < 0,001" if p < 0.001 else f"p = {p:.4f}".replace(".", ",")

def ic_prop(k, n, alpha=ALPHA):
    """IC de una proporción con la aproximación normal vista en clase: p ± Z * SE, con SE = raíz(p(1 - p) / n)."""
    p = k / n
    z = stats.norm.ppf(1 - alpha / 2)
    se = np.sqrt(p * (1 - p) / n)
    return max(0.0, p - z * se), min(1.0, p + z * se)

def welch(a, b, alpha=ALPHA):
    """Prueba t de Welch (no exige varianzas iguales) con el IC de la diferencia de medias, como en el caso de negocio."""
    res = stats.ttest_ind(a, b, equal_var=False)
    ic = res.confidence_interval(confidence_level=1 - alpha)
    return np.mean(a) - np.mean(b), ic.low, ic.high, res.statistic, res.df, res.pvalue

def mes_es(p):
    """Periodo mensual en texto corto, por ejemplo 'sep 2024'."""
    p = pd.Period(p, "M")
    return f"{MESES[p.month - 1]} {p.year}"
```

### 1.2 Llave del proveedor y anonimización

El nombre del proveedor viene en un solo campo con el formato NOMBRE_ENTIDAD_CÓDIGO. La entidad es la sociedad del Banco que contrata (por ejemplo 10001 o 40048) y el código es el identificador del proveedor en el maestro. Separar estas partes es necesario por dos razones. Primero, porque la llave real del proveedor es la entidad junto con el código y no el nombre. Segundo, porque comparar los nombres es la base para encontrar proveedores duplicados, que es el objetivo específico 3.

```python
def separar_proveedor(nombre):
    """Devuelve (nombre_base, entidad, codigo, formato)."""
    if pd.isna(nombre):
        return (np.nan, np.nan, np.nan, "nulo")
    m = re.match(r"^(.*)_(\d+)_(\d+)$", nombre)
    if m:
        return (m.group(1).strip(), m.group(2), m.group(3), "estándar")
    m = re.match(r"^.*\((.*)_(\d+)_(\d+)\)$", nombre)
    if m:
        return (m.group(1).strip(), m.group(2), m.group(3), "nombre repetido entre paréntesis")
    return (nombre.strip(), np.nan, np.nan, "sin entidad ni código")

FORMAS_LEGALES = r"\b(INC|INCORPORATED|LTD|LTEE|LIMITED|LLC|LLP|LP|L P|ULC|CORP|CORPORATION|CO|COMPANY|SAS|S A S|SA|S A|SAA|SRL|S R L|SPA|LTDA|PLC|GMBH|BV|DE CV|S DE RL|THE)\b"

def normalizar_nombre(nombre):
    """Mayúsculas, sin puntuación y sin formas societarias, para comparar nombres de proveedores."""
    if pd.isna(nombre):
        return np.nan
    n = re.sub(r"[^A-Z0-9 ]", " ", str(nombre).upper())
    n = re.sub(FORMAS_LEGALES, " ", n)
    return re.sub(r"\s+", " ", n).strip()

df = raw_df.copy()
df[["prov_nombre_base", "prov_entidad", "prov_codigo", "prov_formato"]] = pd.DataFrame(
    df["Invoice Supplier Name"].apply(separar_proveedor).tolist(), index=df.index)
df["prov_nombre_norm"] = df["prov_nombre_base"].apply(normalizar_nombre)

# si un nombre sin código coincide con un solo proveedor con código, se le asigna ese código
llaves_por_nombre = (df.dropna(subset=["prov_codigo"])
                       .groupby("prov_nombre_norm")[["prov_entidad", "prov_codigo"]]
                       .agg(lambda s: s.unique().tolist()))
sin_codigo = df["prov_formato"].eq("sin entidad ni código")
for idx in df.index[sin_codigo]:
    nombre = df.at[idx, "prov_nombre_norm"]
    if nombre in llaves_por_nombre.index and len(llaves_por_nombre.at[nombre, "prov_codigo"]) == 1:
        df.at[idx, "prov_entidad"] = llaves_por_nombre.at[nombre, "prov_entidad"][0]
        df.at[idx, "prov_codigo"] = llaves_por_nombre.at[nombre, "prov_codigo"][0]

df["prov_llave"] = df["prov_entidad"] + "_" + df["prov_codigo"]
df["prov_formato"].value_counts(dropna=False)
```

Casi todos los nombres siguen el formato esperado (75.578). Sin embargo, hay 42 facturas sin proveedor, 10 donde el nombre aparece repetido entre paréntesis y una que no tiene entidad ni código. Para esa última usamos una regla de imputación, ya que su nombre normalizado coincide con un único proevedor que sí tiene código, así que le asignamos esa llave. En cambio, las 42 facturas sin proveedor las conservamos, pero quedan por fuera de los análisis por proveedor.

Hay algunas señales que solo se pueden calcullar con los nombres reales. Así que, calculamos estas señales antes de anonimizar y después eliminamos los nombres. En ningún momento elnotebook imprime un nombre real.

```python
# estas señales necesitan los nombres reales, por eso van ates de anonimizar

PATRON_INTRAGRUPO = (r"^(?:THE )?BANK OF NOVA SCOTIA|SCOTIABANK|^SCOTIA (?:GBS|GLOBAL|CAPITAL|WEALTH|SERVICIOS|DEALER|"
                     r"LOYALTY|MCLEOD|MORTGAGE|REALTY|LIFE)|SCOTIAMCLEOD|SCOTIATRUST|BANQUE SCOTIA|TANGERINE BANK|"
                     r"1832 ASSET MANAGEMENT|JARISLOWSKY|MD FINANCIAL MANAGEMENT")
df["prov_intragrupo"] = df["prov_nombre_base"].str.upper().str.contains(PATRON_INTRAGRUPO, regex=True, na=False)

df["solicitante_es_aprobador"] = (df["Requested by Name"].notna()
                                  & (df["Requested by Name"] == df["Original Approver Name"]))

logins_multinombre = df.groupby("User Login")["Requested by Name"].nunique().gt(1).sum()
ids_aprob_multinombre = df.groupby("Original Approver ID")["Original Approver Name"].nunique().gt(1).sum()

variantes_por_llave = df.dropna(subset=["prov_llave"]).groupby("prov_llave")["prov_nombre_base"].nunique()
llaves_por_nombre_norm = df.dropna(subset=["prov_llave"]).groupby("prov_nombre_norm")["prov_llave"].nunique()

print(f"Facturas de proveedores intragrupo:            {df['prov_intragrupo'].sum():,}")
print(f"Facturas donde solicitante = aprobador original: {df['solicitante_es_aprobador'].sum():,}")
print(f"Logins asociados a más de un nombre:            {logins_multinombre}")
print(f"IDs de aprobador asociados a más de un nombre:  {ids_aprob_multinombre}")
print(f"Llaves de proveedor con más de un nombre:       {(variantes_por_llave > 1).sum()}")
print(f"Nombres normalizados con más de una llave:      {(llaves_por_nombre_norm > 1).sum()}")
```

## 2. Cobertura temporal y comparabilidad de periodos

Que cada año tenga fechas de enero a diciembre no es suficiente para compararlos. Lo que hay que revisar, es si las facturas de cada periodo se parecen. Es decir, si entran por los mismos canales, si tienen la misma proporcion de contratos y si tienen los mismos campos informados. Si la composición cambia, una diferencia entre años puede deberse a un cambio en el proceso o en lo que entra al sistema, y no a un cambio en el riesgo.

```python
df["mes"] = df["Invoice Created Date"].dt.to_period("M")
df["anio"] = df["Invoice Created Date"].dt.year

mensual = df.groupby("mes").agg(facturas=(MONTO, "size"))
pos = np.arange(len(mensual))
pos_ruptura = list(mensual.index).index(pd.Period("2024-08", "M"))
fig, ax = plt.subplots(figsize=(16, 4.5))
ax.bar(pos, mensual["facturas"], color="steelblue", width=0.85)
ax.axvline(pos_ruptura - 0.5, color="crimson", ls="--", lw=1.5)
ax.text(pos_ruptura - 0.3, mensual["facturas"].max() * 0.95, " agosto de 2024: cambio de régimen", color="crimson")
ax.set_title("Número de facturas creadas por mes (todas las facturas ≥ 10.000 CAD)")
ax.set_xlabel("Mes de creación"); ax.set_ylabel("Número de facturas")
ax.set_xticks(pos[::3]); ax.set_xticklabels([mes_es(p) for p in mensual.index[::3]], rotation=45)
plt.tight_layout(); plt.show()
antes = mensual.loc[mensual.index < pd.Period("2024-08", "M"), "facturas"]
despues = mensual.loc[mensual.index >= pd.Period("2024-09", "M"), "facturas"]
print(f"Antes de agosto de 2024: entre {antes.min()} y {antes.max()} facturas por mes (mediana {antes.median():.0f})")
print(f"Desde septiembre de 2024: entre {despues.min()} y {despues.max()} facturas por mes (mediana {despues.median():.0f})")
print("Julio, agosto y septiembre de 2024:", mensual.loc["2024-07":"2024-09", "facturas"].tolist())

df.groupby("anio")["Invoice Created Date"].agg(["min", "max", "count"]).assign(
    min=lambda t: t["min"].dt.strftime("%d/%m/%Y"), max=lambda t: t["max"].dt.strftime("%d/%m/%Y"))
```

Hasta julio de 2024 el volumen se mueve entre 542 y 868 facturas por mes (mediana de 654), y en agosto de 2024 salta: pasa de 845 factturas en julio a 1.538 en agosto y 2.397 en septiembre. Lo que muestra el mapa de métodos de creacion, es de dónde viene el cambio. Desde agosto de 2024 el metodo buyer_load (facturas que el Banco carga por sftp) pasa de 4,1% a 29,3% de las facturas, y enterprise_ui (facturas que digita el personal interno) pasa de 1,0% a 14,5%. Al mismo tiempo, inbound_invoice, que son las facturas que llegan por correo, baja de 18,9% a 4,2%. Por lo tanto, lo más probable es que desde agosto de 2024 empezaran a entrar a SmartBuy facturas sin orden de compra que antes se procesaban por fuera de la herramienta. Esta hipótesis hay que validarla con el stakeholder.

```python
def cramers_v(tabla):
    chi2 = stats.chi2_contingency(tabla)[0]
    n = tabla.to_numpy().sum()
    r, k = tabla.shape
    return np.sqrt(chi2 / (n * (min(r, k) - 1)))

df["regimen"] = np.where(df["Invoice Created Date"] < "2024-08-01", "anterior (junio 2022 a julio 2024)", "actual (agosto 2024 a julio 2026)")
resultados = []
for var, serie in {"canal": df["Invoice Channel"],
                   "tiene contrato": df["Contract Number"].notna(),
                   "categoría informada": df["Commodity Level 1"].notna(),
                   "entidad del proveedor": df["prov_entidad"].fillna("sin dato")}.items():
    tabla = pd.crosstab(df["regimen"], serie)
    chi2, p, dof, _ = stats.chi2_contingency(tabla)
    resultados.append({"variable": var, "chi2": chi2, "gl": dof, "p_valor": p, "V_de_Cramer": cramers_v(tabla)})

ant = df.loc[df["regimen"].str.startswith("anterior"), MONTO].dropna()
act = df.loc[df["regimen"].str.startswith("actual"), MONTO].dropna()
dif, lo, hi, t, gl, p = welch(np.log10(ant), np.log10(act))
print(f"Mediana del monto: anterior {ant.median():,.0f} CAD | actual {act.median():,.0f} CAD")
print(f"Diferencia de medias en log10 (anterior menos actual): {dif:.3f}, IC 95% [{lo:.3f}, {hi:.3f}] | "
      f"t de Welch = {t:.1f}, gl = {gl:,.0f}, {fmt_p(p)} | en escala original el monto típico anterior es {10**dif:.2f} veces el actual")
pd.DataFrame(resultados).round(4)
```

Las pruebas chi cuadrado confirman que la composición del regimen anterior (junio de 2022 a julio de 2024) y ladel actual (agosto de 2024 a julio de 2026) es distinta. La asociación es moderada para el canal (V de Cramér de 0,36) y para tener contrato (0,39), y está entre débil y moderada para la categoria informada (0,23) y la entidad del proveedor (0,20). Además, la mediana del monto también cambia, de 36.449 CAD antes a 30.849 CAD después. Dado que, el monto es muy asimétrico, hicimos la comparación en escala log10 con la prueba t de Welch, que no exige varianzas iguales. La diferencia de medias en log10 es 0,062 (IC 95% de 0,052 a 0,071; p < 0,001), es decir, el monto típico del régimen anterior era unas 1,15 veces el actual. Con 75 mil facturas casi cualquier diefrencia sale significativa, así que aquí lo que importa es el tamaño de la V de Cramér, y en canal y contrato no es pequeño.

Dado que, la factura está anulada, no afecta el gasto aprobado. Sin embargo, sí distorsionaria cualquier estadística de las facturas anuladas o de los montos en general, como se ve enla siguiente celda.

La revisión no se puede quedar solo en el caso más grande. Por eso, definimos una regla que se aplica a todas las facturas por encima del percentil 99,9. Un extremo se considera verificado por recurrencia si el mismo proveedor factura montos parecidos (más o menos 15%) al menos otras dos veces. Por ejemplo, esto pasa con las licencias o los servicios recurrentes de gran tamaño, donde el monto es alto pero es el normal para ese proveedor. Si no hay esa recurrencia, el extremo qudea como no verificado. En ese caso, lo excluimos de las correlaciones y las pruebas, lo dejamos en el gasto total si fue aprobado y lo llevamos a validación con el Banco.

```python
# extremos por encima del p99,9:
#  error_monto: monto imposible para la moneda, pasa a nulo
#  extremo verificado: el proveedor factura montos parecidos (±15%) otras 2 veces o más, se deja
#  extremo_no_verificado: no se usa en correlaciones ni pruebas, sí en el gasto total
p999 = df[MONTO].quantile(0.999)
df["error_monto"] = False
df.loc[caso.name, "error_monto"] = True

def montos_similares(fila, tolerancia=0.15):
    otros = df.loc[(df["proveedor_id"] == fila["proveedor_id"]) & (df.index != fila.name), "monto_cad"]
    return int(((otros - fila["monto_cad"]).abs() <= tolerancia * fila["monto_cad"]).sum())

extremos = df[(df["monto_cad"] > p999) & ~df["error_monto"]].copy()
extremos["montos_similares_del_proveedor"] = extremos.apply(montos_similares, axis=1)
extremos["verificado_por_recurrencia"] = extremos["montos_similares_del_proveedor"] >= 2
df["extremo_no_verificado"] = df.index.isin(extremos.index[~extremos["verificado_por_recurrencia"]])

print(f"Percentil 99,9 del monto: {p999:,.0f} CAD | facturas por encima (sin el error): {len(extremos)}")
print("Verificadas por recurrencia:", extremos["verificado_por_recurrencia"].sum(),
      "| no verificadas:", (~extremos["verificado_por_recurrencia"]).sum())
print(extremos.loc[~extremos["verificado_por_recurrencia"],
                   ["proveedor_id", "Invoice Status", "monto_cad", "veces_mediana_proveedor",
                    "montos_similares_del_proveedor"]].sort_values("monto_cad", ascending=False))

anuladas = df[df["Invoice Status"] == "voided"]
print(f"\nFacturas anuladas, media con el error: {anuladas['monto_cad'].mean():,.0f} CAD | "
      f"sin el error: {anuladas.loc[~anuladas['error_monto'], 'monto_cad'].mean():,.0f} CAD | "
      f"suma con el error: {anuladas['monto_cad'].sum() / 1e6:,.0f} M | sin el error: "
      f"{anuladas.loc[~anuladas['error_monto'], 'monto_cad'].sum() / 1e6:,.0f} M")

# el monto con error pasa a nulo, la factura no se borra
df.loc[df["error_monto"], "monto_cad"] = np.nan
```

El percentil 99,9 es 17.527.986 CAD y, sin contar el error, hay 75 facturas por encima. De elas, 57 están verificadas por recurrencia y 18 no. Entre las 18 no verificadas hay borradores y anuladas, como el borrador de 660 millones de CAD, y también algunas aprobadas, como una de 134,96 millones de CAD que es 2.742 veces la mediana de su proveedor. Ninguna la declaramos error porque no hay evidencia para hacerlo, pero tampoco enttran a las correlaciones sin verificar.

La última parte de la salida muestra por qué era importante tratar el error. Con la factura de 1.070 milones, lamedia de las facturas anuladas es 544.451 CAD y su suma es 2.246 millones.
```python
prov = (df.dropna(subset=["proveedor_id"]).groupby("proveedor_id")
          .agg(nombre=("prov_nombre_norm", "first"), entidad=("prov_entidad", "first"),
               facturas=("prov_nombre_norm", "size")).reset_index())
pares = []
for entidad, g in prov.groupby("entidad"):
    g = g.sort_values("nombre").reset_index(drop=True)
    nombres_g = g["nombre"].tolist()
    for i in range(len(g)):
        for j in range(i + 1, len(g)):
            if nombres_g[j][:3] != nombres_g[i][:3]:   # solo nombres con las mismas 3 primeras letras, si no tarda muchísimo
                break
            r = ratio(nombres_g[i], nombres_g[j])
            if r >= 0.90:
                pares.append({"entidad": entidad, "prov_a": g.at[i, "proveedor_id"], "prov_b": g.at[j, "proveedor_id"],
                              "similitud": round(r, 3), "identico": nombres_g[i] == nombres_g[j]})
pares = pd.DataFrame(pares)
print("Pares de proveedores con similitud de nombre >= 0,90 en la misma entidad:", len(pares))
print("…nombre normalizado idéntico:", pares["identico"].sum(), "| casi idéntico (0,90 a 0,99):", (~pares["identico"]).sum())
pares.sort_values("similitud").head(8)
```

Usando la similitud de Levenshtein, como en el tutorial de calidad, aparecen 56 pares de proveedores en la misma entidad con una similtud de 0,90 o más. De ellos, 36 tenen el nombre normalizado idéntico y 20 son casi idénticos (similitud entre 0,90 y 0,99). Estos 20 son justamente los que una busqueda exacta no encontraría.

Con esto creamos grupo_nombre_id para consolidar el gasto por nombre normalizado. Las llaves no se fusionan automáticamente, dado quedos empresas distintas pueden tener nombres parecidos (homónimos). Así que, le entregamos la lista de pares al stakeholder para que confirme cuáles son la misma empresa.

```python
x = aprobadas["monto_cad"]
q1, q3 = x.quantile([.25, .75])
log_x = np.log10(x)
lq1, lq3 = log_x.quantile([.25, .75]); liqr = lq3 - lq1
estad = pd.Series({
    "facturas": len(x), "media": x.mean(), "mediana": x.median(), "desv_estandar": x.std(),
    "IQR": q3 - q1, "p5": x.quantile(.05), "Q1": q1, "Q3": q3,
    "p95": x.quantile(.95), "p99": x.quantile(.99), "máximo": x.max(),
    "asimetria": x.skew(), "curtosis": x.kurtosis(),
    "asimetria_log10": log_x.skew(),
    "pct_atipicos_IQR_escala_original": ((x > q3 + 1.5 * (q3 - q1)).mean() * 100),
    "pct_atipicos_IQR_escala_log": (((log_x > lq3 + 1.5 * liqr) | (log_x < lq1 - 1.5 * liqr)).mean() * 100),
})
estad.to_frame("monto_cad (facturas aprobadas)")
```

## 5. Análisis bivariado y multivariado

### 5.1 Variables objetivo provisionales

El Banco no tiene registrado cuáles facturas fueron pagos duplicados. Por eso, mientras conseguimos una muestra etiquetada usamos dos variables provisionales, cada una con sus limitaciones.

La primera es la fricción (con_friccion), que indica si la factura terminó anulada, disputada o rechazada. La calculammos sobre las facturas resueltas y es una señal de que algo en la factura no salió bien. Sin ebargo, no mide fraude ni duplicados, ya que como vimos en la sección 3.5 muchas anulaciones son correcciones normales.

La segunda es el candidato a duplicado (candidato_duplicado). Este consiste en una factura aprobada que tiene otra aprobada del mismo proveedor con el mismo monto en los 7 días anteriores, o que comparte un número de factura casi idéntico con otra factura aprobada. Las coincidencias con facturas anuladas no cuentan, porque son correcciones. Es un filtro amplio que también incluye la facturación en lote normal, y la usamos para describir en dónde se concentra el riesgo.

```python
print("Tasa global de fricción en facturas resueltas:", f"{resueltas['con_friccion'].mean():.2%}")
print(resueltas["Invoice Status"].value_counts().to_frame("facturas").T)

def tasa_con_ic(g):
    k, n = g.sum(), g.count()
    lo, hi = ic_prop(k, n)
    return pd.Series({"facturas": n, "con_friccion": k, "tasa_pct": k / n * 100, "ic95_inf": lo * 100, "ic95_sup": hi * 100})

ap_v = ap[(ap["Invoice Created Date"] >= VENTANA_INI) & (ap["Invoice Created Date"] < VENTANA_FIN)].copy()
ap_v["dup_mismo_monto_7d"] = ap_v["dias_desde_mismo_monto"].between(0, 7)
ap_v["dup_numero"] = ap_v.index.isin(colision_aprob_idx)
ap_v["candidato_duplicado"] = ap_v["dup_mismo_monto_7d"] | ap_v["dup_numero"]
print(f"\nAprobadas en la ventana: {len(ap_v):,} | candidatas a duplicado: {ap_v['candidato_duplicado'].sum():,} "
      f"({ap_v['candidato_duplicado'].mean():.1%})")
```

La tasa global de fricción es 5,33%, es decir, hay 2.436 anuladas, 186 disputadas y 108 rechazadas frente a 48.457 aprobadas. Entre las aprobadas de la ventana, 3.078 (el 6,4%) son candidatas a duplicado. Las dos variables están desbalanceadas, así que desde ya sabemos que el modelo no se puede evaluar con el porcentaje total de aciertos. Por ejemplo, si marcara todo como normal acertaría más del 93% de las veeces sin encontrar ningún duplicado. Lo que sivre es mirar qué proporción de los casos marcados se confirma y qué proporción de los duplicados reales logra encontrar.

### 5.2 H1: ¿dónde aparecen los posibles duplicados?

Hipótesis

H0: la proporción de facturas aprobadas con el mismo monto en 0 a 7 días es igual en todos los canales.

H1: al menos un canal tiene una proporción distinta.

Como las dos variables son categoricas, uamos la prueba chi cuadrado de independencia con un nivel de significancia de 0,05.

```python
t_dup = ap_v.groupby("Invoice Channel").agg(
    facturas=("monto_cad", "size"),
    pct_mismo_monto_0_a_7_dias=("dup_mismo_monto_7d", lambda s: s.mean() * 100),
    pct_numero_casi_identico=("dup_numero", lambda s: s.mean() * 100))
tabla_d = pd.crosstab(ap_v["Invoice Channel"], ap_v["dup_mismo_monto_7d"])
chi2_d, p_d, _, _ = stats.chi2_contingency(tabla_d)
print(f"Canal vs mismo monto en 0 a 7 días: chi2 = {chi2_d:,.1f}, {fmt_p(p_d)}, V de Cramér = {cramers_v(tabla_d):.3f}")
print(f"Candidatos en la ventana: mismo monto en 0 a 7 días = {ap_v['dup_mismo_monto_7d'].sum():,} | "
      f"número casi idéntico = {ap_v['dup_numero'].sum():,} | cualquiera = {ap_v['candidato_duplicado'].sum():,} "
      f"({ap_v['candidato_duplicado'].mean():.1%} de las aprobadas)")
t_dup.sort_values("pct_mismo_monto_0_a_7_dias", ascending=False).round(2)
```

Rechazamos H0 (chi² = 993,8; p < 0,001) y la asociación está entre débil y moderada (V de Cramér de 0,143). El portal de proveedores (csp) tiene la proporcion más alta de facturas con el mismo monto en 0 a 7 días, con el 10,16% de sus facturas aprobadas, frente al 4,53% de sftp y el 1,23% de eamil. Esto tiene sentido, dado que el portal es el canal donde el proveedor carga sus propias facturas y a veces carga varias a la vez. En cambio, el numero casi identico a otra factura aprobada es mucho más raro (32 facturas en la ventana) y aparece en todos los canales menos en cxml.

Para el negocio, esto quiere decir que la mayoria de los candidatos viene del portal, así que ese canal es la primera prioridad para el modelo. Al mismo tiempo, es el canal donde más hay que separar la facturación en lote normal de los duplicados.

### 5.3 H2: canal de ingreso, cadena de aprobación y riesgo

a) Canal vs fricción

H0: la tasa de fricción es la misma en todos los canales.

H1: al menos un canal tiiene una tasa distinta.

Usamos la prueba chi cuadrado, y reportamos las tasas con intervalos de confianza al 95% calculados con la proporción más o menos Z veces su error estándar. Además, verificamos que tdas las frecuencias esperadas fueran mayores a 5 (la mínima es 78).

```python
t_canal = resueltas.groupby("Invoice Channel")["con_friccion"].apply(tasa_con_ic).unstack()
t_canal = t_canal[t_canal["facturas"] >= 30].sort_values("tasa_pct", ascending=False)
tabla = pd.crosstab(resueltas["Invoice Channel"], resueltas["con_friccion"]).loc[t_canal.index]
chi2, p, dof, esperado = stats.chi2_contingency(tabla)
print(f"Chi cuadrado = {chi2:,.1f}, gl = {dof}, {fmt_p(p)}, V de Cramér = {cramers_v(tabla):.3f}; "
      f"frecuencia esperada mínima = {esperado.min():.0f}")

fig, ax = plt.subplots(figsize=(10, 4))
ax.errorbar(t_canal.index, t_canal["tasa_pct"],
            yerr=[t_canal["tasa_pct"] - t_canal["ic95_inf"], t_canal["ic95_sup"] - t_canal["tasa_pct"]],
            fmt="o", color="steelblue", capsize=5)
ax.axhline(resueltas["con_friccion"].mean() * 100, color="gray", ls="--", lw=1, label="tasa global")
ax.set_title("Tasa de fricción (anulada, disputada o rechazada) por canal, con IC 95%")
ax.set_xlabel("Canal de ingreso"); ax.set_ylabel("% de facturas con fricción"); ax.legend()
plt.tight_layout(); plt.show()
t_canal.round(2)
```

Rechazamos H0 (chi² = 212,0; p < 0,001), pero la asociación es debil (V de Cramér de 0,064). Las tasas de fricción por canal son:

- sftp (tasa de fricción: 6,32%; ic 95%: 5,93% a 6,72%)
- csp (tasa de fricción: 6,11%; ic 95%: 5,79% a 6,42%)
- cxml (tasa de friccion: 4,78%; ic 95%: 3,69% a 5,87%)
- ui (tasa de fricción: 3,26%; ic 95%: 2,92% a 3,60%)
- email (tasa de fricción: 1,49%; ic 95%: 0,98% a 1,99%)

Los intervalos de sftp y csp se cruzan, así que entre esos dos canales no hay evidencia de una diferencia. Por lo tanto, la diferencia real está entre ese par y los canales internos (ui y email).

Validación de supuestos: independencia

Las pruebas anteriores suponen que cada factura es independiente, pero las facturas de un mismo proveedor comparten precios, procesos y errores. Para corregirlo, recalculamos el intervalo de la diferencia más grande con un bootstrap por conglomerados, es decir, remuestreando proveedores y no facturas.

```python
# las facturas de un mismo proveedor no son independientes,
# por eso el bootstrap remuestrea proveedores y no facturas
canal_alto, canal_bajo = t_canal.index[0], t_canal.index[-1]
sub = resueltas[resueltas["Invoice Channel"].isin([canal_alto, canal_bajo])]
por_prov = sub.groupby(["proveedor_id", "Invoice Channel"])["con_friccion"].agg(["sum", "count"]).unstack(fill_value=0)
rng = np.random.default_rng(RANDOM_STATE)
difs = []
idx = np.arange(len(por_prov))
for _ in range(2000):
    m = por_prov.iloc[rng.choice(idx, size=len(idx), replace=True)].sum()
    difs.append(m[("sum", canal_alto)] / m[("count", canal_alto)] - m[("sum", canal_bajo)] / m[("count", canal_bajo)])
dif_obs = t_canal.loc[canal_alto, "tasa_pct"] - t_canal.loc[canal_bajo, "tasa_pct"]
lo, hi = np.percentile(difs, [2.5, 97.5]) * 100
n_alto, n_bajo = t_canal.loc[canal_alto, "facturas"], t_canal.loc[canal_bajo, "facturas"]
p_a, p_b = t_canal.loc[canal_alto, "tasa_pct"] / 100, t_canal.loc[canal_bajo, "tasa_pct"] / 100
se_ingenuo = np.sqrt(p_a * (1 - p_a) / n_alto + p_b * (1 - p_b) / n_bajo) * 100
print(f"Diferencia {canal_alto} - {canal_bajo}: {dif_obs:.2f} puntos porcentuales")
print(f"IC 95% ingenuo (asume independencia):     [{dif_obs - 1.96 * se_ingenuo:.2f}, {dif_obs + 1.96 * se_ingenuo:.2f}]")
print(f"IC 95% bootstrap por proveedor (2.000 rep.): [{lo:.2f}, {hi:.2f}]")
```

La diferencia entre sftp y email es de 4,84 puntos porcentuales. El intervalo que supone independencia va de 4,20 a 5,48, mientras que el intervalo por conglomerados va de 3,52 a 5,75, casi el doble de ancho. La diferencia se mantiene porque el intervalo no incluye el 0. Sin embargo, si ignoramos que las facturas se agrupan por proveedor, obtenemos una precisión que no es real. Es por esto, que en la regresión de la sección 5.6 hicimos lo mismo, con errores estándar agrupados por proveedor.

b) Cadena de aprobación vs fricción

Antes de usar los campos de aprobacion como variables explicativas, hay que revisar si el dato existe antes de conocer el resultado de la factura.

```python
fuga = base.groupby("Invoice Status").agg(
    facturas=("monto_cad", "size"),
    pct_sin_n_aprobadores=("n_aprobadores", lambda s: s.isna().mean() * 100),
    pct_sin_aprobador_original=("Original Approver Name", lambda s: s.isna().mean() * 100))
sin_cad = base[(base["Invoice Status"] == "approved") & base["n_aprobadores"].isna()]
print(f"Aprobadas sin cadena de aprobación registrada: {len(sin_cad):,} de {(base['Invoice Status'] == 'approved').sum():,}")
print("Canal de esas facturas (%):", (sin_cad["Invoice Channel"].value_counts(normalize=True) * 100).round(1).to_dict())
fuga.loc[["approved", "voided", "disputed", "rejected"]].round(1)
```

Esta tabla muestra uno de los hallazgos más importantes del notebook. El numero de aprobadores está vacío en el 99,3% de las facturas anuladas y en el 50,5% de las aprobadas. Una factura anulada casi nunca llega a tener cadena de aprobacion porque se anula antes. Si se usara Number of Approvers para explicar la friccion, la relación saldría casi perfecta, pero sería flsa, dado que el dato se llena como consecuencia del resultado. Esto es una fuga de informacion (*data leakage*).

Este resultado también sirve para revisar una justificación del punto 1. Allí dijimos que trabajamos con las facturas de 10.000 CAD o más porque desde ese monto requieren al menos un aprobador y una doble revisión de Procurement. Los datos lo confirman solo en parte, ya que 24.454 de las 48.457 aprobadas (el 50,5%) no tienen una cadena de aprobacion registrada en la factura. De esas, el 78,3% entro por el portal (csp) y el 5,6% por cxml, que se emiten contra una orden de compra que ya pasó por aprobación (sección 3.1). Así que, la doble revisión puede estar ocurriendo en la orden de compra y no en la factura. Esto no cambia el análisis, pero sí cómo presentamos el corte de 10.000 CAD. Es decir, lo tomamos como el filtro de la extracción, y la regla de aprobación la vamos a validar con el Banco.

Como resultado, decidimos no usar los campos de aprobación para explicar la fricción ni como variables del modelo cuando la etiqueta dependa del estado final. La parte de la hipótesis 2 sobre la cadena de aprobación la replanteamos sobre las facturas aprobadas, que es donde la cadena sí existe, usando como resultado el candidato a duplicado.

c) Cadena de aprobación vs candidatos a duplicado (facturas aprobadas)

H0: la proporción de candidatos a duplicado no depende del número de aprobadores ni de que el solicitante sea el aprobador.

H1: sí depende.

```python
ap_v["aprobadores_cat"] = (pd.cut(ap_v["n_aprobadores"], [0, 1, 2, 4, 100], labels=["1", "2", "3 a 4", "5 o más"])
                           .cat.add_categories("sin cadena").fillna("sin cadena").astype(str))

def tasa_con_ic_col(g, col):
    k, n = g[col].sum(), len(g)
    lo, hi = ic_prop(k, n)
    return pd.Series({"facturas": n, "candidatas": k, "tasa_pct": k / n * 100, "ic95_inf": lo * 100, "ic95_sup": hi * 100})

t_ac = ap_v.groupby("aprobadores_cat").apply(tasa_con_ic_col, col="candidato_duplicado").loc[["sin cadena", "1", "2", "3 a 4", "5 o más"]]
tabla_ac = pd.crosstab(ap_v["aprobadores_cat"], ap_v["candidato_duplicado"])
chi2_ac, p_ac, _, _ = stats.chi2_contingency(tabla_ac)
print(f"Aprobadores vs candidato a duplicado: chi2 = {chi2_ac:,.1f}, {fmt_p(p_ac)}, V de Cramér = {cramers_v(tabla_ac):.3f}")
print(t_ac.round(2), "\n")

sub_c = ap_v[ap_v["Invoice Channel"].isin(["sftp", "ui", "email"])]
t_ac2 = sub_c.groupby(sub_c["n_aprobadores"].isna().map({True: "sin cadena", False: "con cadena"})).apply(
    tasa_con_ic_col, col="candidato_duplicado")
print("Solo canales sftp, ui y email:"); print(t_ac2.round(2), "\n")

con_cadena = ap_v[ap_v["n_aprobadores"].notna()]
t_sod = con_cadena.groupby("solicitante_es_aprobador").apply(tasa_con_ic_col, col="candidato_duplicado")
tabla_s = pd.crosstab(con_cadena["solicitante_es_aprobador"], con_cadena["candidato_duplicado"])
chi2_s, p_s, _, _ = stats.chi2_contingency(tabla_s)
print(f"Solicitante = aprobador vs candidato a duplicado: chi2 = {chi2_s:,.1f}, {fmt_p(p_s)}, V de Cramér = {cramers_v(tabla_s):.3f}")
print(t_sod.round(2))

rho_ma, p_ma = stats.spearmanr(con_cadena["monto_cad"], con_cadena["n_aprobadores"])
print(f"\nSpearman monto vs número de aprobadores (aprobadas con cadena): rho = {rho_ma:.3f}, {fmt_p(p_ma)}")
```

Con el número de aprobadores rechazamos H0 (chi² = 878,1; p < 0,001; V de Cramér de 0,135). Sin embargo, esta comparación global está muy influida por el canal, porque las facturas sin cadena son sobre todo del portal, que es donde más hay facturación en lote. Para aislar ese efecto, repetimos la comparación solo con sftp, ui y email, donde la cadena de aprobación es lo normal. En ese caso, las facturas sin cadena tienen 5,02% de candidatos frente a 3,22% de las que sí tienen cadena, y los intervalos no se cruzan. Esto apoya la hipótesis 2, ya que las facturas que no pasan por una cadena de aprobación tienen más candidatos a duplicado incluso dentro de los mismos canales.

Cuando el solicitante es el aprobador, la tasa de candidatos es 3,77% frente a 2,90%. La diferencia es significativa (p = 0,0011) pero la asociación es muy pequeña (V de 0,021), así que es una señal complementaria y no un criterio fuerte por sí sola. Por último, la correlación de Spearman entre el monto y el número de aprobadores es de 0,323. Es decir, a mayor monto, más aprobadores, lo cuál es coherente con una política de aprobacion por niveles de monto.

```python
d = con_cadena.assign(aprobadores=con_cadena["n_aprobadores"].clip(upper=8).astype(int).astype(str).replace("8", "8+"),
                      log_monto=np.log10(con_cadena["monto_cad"]))
orden_a = ["1", "2", "3", "4", "5", "6", "7", "8+"]
fig, ax = plt.subplots(figsize=(10, 4))
sns.boxplot(data=d, x="aprobadores", y="log_monto", order=orden_a, showfliers=False, ax=ax, color="steelblue")
ax.set_title("Monto aprobado según el número de aprobadores (escala log10)")
ax.set_xlabel("Número de aprobadores"); ax.set_ylabel("log10(monto en CAD)")
plt.tight_layout(); plt.show()
print(d.groupby("aprobadores")["monto_cad"].median().loc[orden_a].round(0).to_frame("mediana_monto_cad").T)
an_a = stats.f_oneway(*[d.loc[d["aprobadores"] == a, "log_monto"] for a in orden_a])
print(f"ANOVA de una vía (log10 del monto por número de aprobadores): F = {an_a.statistic:,.1f}, {fmt_p(an_a.pvalue)}")
```

La grafica mustra que el número de aprobadores sigue al monto, aunque no de forma perfecta. ### 5.4 H3: ¿qué proveedores concentran el riesgo?

Este análisis lo hacemos a nivel de proveedor porque esa es la unidad independiente. Usamos los 669 proveedores que tienen al menos 10 facturas resueltas en la ventana, para que su tasa de fricción no dependa de una o dos facturas. Dado que, las variables son asimétricas y la relacion no tiene por qué ser lineal, usamos Spearman.

H0: ρ = 0 entre cada característica del proveedor y su tasa de fricción.

H1: ρ ≠ 0.

```python
primera = df.groupby("proveedor_id")["Invoice Created Date"].min()   # historia completa desde jun-2022
prov_df = resueltas.groupby("proveedor_id").agg(
    facturas=("con_friccion", "size"), tasa_friccion=("con_friccion", "mean"),
    gasto_aprobado=("monto_cad", lambda s: s[resueltas.loc[s.index, "con_friccion"] == 0].sum()),
    mediana_monto=("monto_cad", "median"),
    pct_sin_contrato=("tiene_contrato", lambda s: 1 - s.mean()),
    intragrupo=("prov_intragrupo", "max"))
prov_df["antiguedad_dias"] = np.clip((pd.Timestamp(VENTANA_INI) - prov_df.index.map(primera)).days, 0, None)
prov_df["nuevo_en_ventana"] = prov_df["antiguedad_dias"] == 0
prov10 = prov_df[prov_df["facturas"] >= 10].copy()
print("Proveedores con >= 10 facturas resueltas:", len(prov10))

filas = []
for v in ["facturas", "gasto_aprobado", "mediana_monto", "antiguedad_dias", "pct_sin_contrato"]:
    rho, p = stats.spearmanr(prov10[v], prov10["tasa_friccion"])
    filas.append({"variable": v, "rho_spearman": rho, "p_valor": p})
print(pd.DataFrame(filas).round(4), "\n")

t_nuevo = prov10.groupby("nuevo_en_ventana")["tasa_friccion"].describe()[["count", "mean", "50%"]]
dif, lo, hi, t, gl, p = welch(prov10.loc[prov10["nuevo_en_ventana"], "tasa_friccion"] * 100, prov10.loc[~prov10["nuevo_en_ventana"], "tasa_friccion"] * 100)
print(t_nuevo.round(3))
print(f"t de Welch, nuevos menos antiguos: {dif:.2f} puntos porcentuales, IC 95% [{lo:.2f}, {hi:.2f}], {fmt_p(p)}")
t_ig = prov10.groupby("intragrupo")["tasa_friccion"].describe()[["count", "mean", "50%"]]
print(t_ig.round(3))

fig, ax = plt.subplots(figsize=(10, 4.5))
ax.scatter(prov10["facturas"], prov10["tasa_friccion"] * 100, s=12, alpha=.5, color="steelblue")
ax.set_xscale("log")
ax.set_title("Tasa de fricción por proveedor vs número de facturas (proveedores con ≥ 10 facturas)")
ax.set_xlabel("Facturas resueltas del proveedor en la ventana (escala log)"); ax.set_ylabel("% de facturas con fricción")
plt.tight_layout(); plt.show()
```

La hipótesis 3 no se apoya en la dirección que planteamos. Esperábamos que los proveedores nuevos o con poca historia tuvieran más anomalias, pero pasa lo contrario. Los proveedores cuya primera factura cae dentro de la ventana tienen una tasa media de fricción de 5% (mediana de 0%), frente a 7% (mediana de 4%) de los que ya tenían historia. Con la prueba t de Welch la diferencia es de 2,19 puntos porcentuales menos para los nuevos (IC 95% de 0,62 a 3,76 puntos; p = 0,0063), y la antigüedad tiene una asociación positiva pero debil con la fricción (ρ = 0,18). Una posible explicación es que los proveedores antiguos son los que más usan el portal y más corrigen facturas. Además, la antigüedad está truncada porque los datos empiezan enjunio de 2022.

Los proveedores con más facturas y más gato tienen un poco más de friccion (ρ de 0,14 y 0,15), aunque son asociaciones débiles. En cambio, los proveedores con más facturas sin contrato tienen menos fricción (ρ negativo de 0,28), lo cuál es coherennte con que la fricción se concentre en el flujo con orden de compra del portal. Los proveedores intragrupo tienen una tasa media más alta (11% frente a 6%), pero son solo 29, así que en la seccion 5.6 se revisa si esa diferencia se mantiene al controlar otras variables. Por ultimo, en el diagrama de dispersion vemos que las tasas más altas (más de 40%) aparecen en proveedores con pocas facturas, donde una o dos anulaciones mueven mucho el porcentaje.

### 5.5 H4: ¿las facturas con un monto atípico para su proveedor tienen más fricción?

En el punto 4 planteamos la hipótesis con un ejemplo: las facturas con más de tres desviaciones sobre la mediana de su proveedor. Sin embargo, esa regla tiene un problema con estos datos. Con pocas facturas, una factura no puede quedar a tres desviaciones estándar de la media de su grupo, porque el máximo posible es (n menos 1) dividido entre la raíz de n, y ese valor solo pasa de 3 cuando el proveedor tiene 11 facturas o más. En estos datos son 364 de los 985 proveedores analizados (el 37,0%). Es decir, con esa regla más de un tercio de los proveedores nunca tendría una factura atípica. Además, la pregunta del punto 4 tambien mencionaba la categoría, pero el 35,5% de las facturas no tiene categoría, así que la referencia más confiable es el propio proveedor.

Por lo tanto, usamos la regla del rango intercuartílico, pero calculada para cada proveedor sobre el log10 del monto. Es decir, la cerca la calculamos por grupo para no sesgar la comparación. Trabajamos con los proveedores que tienen al menos 5 facturas resueltas y un IQR mayor a cero, sin los extremos no verificados.

Cada factura queda en uno de cinco niveles: muy bajo (por debajo de Q1 menos 1,5 IQR de su proveedor), bajo (entre esa cerca y Q1), habitual (entre Q1 y Q3), alto (entre Q3 y Q3 más 1,5 IQR) y muy alto (por encima de la cerca superior).

H0: la tasa de fricción es igual para las facturas por encima de la cerca superior y para las que estan dentro de las cercas.

H1: la tasa es distinta.

```python
# IQR por proveedor sobre el log10 del monto,
# para no comparar proveedores grandes con pequeños; solo proveedores con 5+ faturas
r = resueltas[~resueltas["extremo_no_verificado"]].copy()
r["log_monto"] = np.log10(r["monto_cad"])
g = r.groupby("proveedor_id")["log_monto"]
r["n_prov"] = g.transform("size")
r["q1_prov"] = g.transform(lambda s: s.quantile(.25))
r["q3_prov"] = g.transform(lambda s: s.quantile(.75))
r["mediana_log_prov"] = g.transform("median")
r["iqr_prov"] = r["q3_prov"] - r["q1_prov"]
r = r[(r["n_prov"] >= 5) & (r["iqr_prov"] > 0)].copy()
r["desvio_log"] = r["log_monto"] - r["mediana_log_prov"]
cerca_inf = r["q1_prov"] - 1.5 * r["iqr_prov"]; cerca_sup = r["q3_prov"] + 1.5 * r["iqr_prov"]
r["nivel_monto"] = np.select(
    [r["log_monto"] < cerca_inf, r["log_monto"] < r["q1_prov"], r["log_monto"] <= r["q3_prov"], r["log_monto"] <= cerca_sup],
    ["muy bajo", "bajo", "habitual", "alto"], default="muy alto")
orden_n = ["muy bajo", "bajo", "habitual", "alto", "muy alto"]
t_z = r.groupby("nivel_monto")["con_friccion"].apply(tasa_con_ic).unstack().loc[orden_n]
print("Facturas con historia suficiente:", len(r), "| proveedores:", r["proveedor_id"].nunique())
# con n facturas el z no puede pasar de (n-1)/raíz(n): la regla de 3 desviaciones
# solo funciona con 11 facturas o más, por eso no se usa
prov_n = r.groupby("proveedor_id")["n_prov"].first()
print(f"Proveedores con menos de 11 facturas, donde 3 desviaciones estándar no se pueden alcanzar: {(prov_n < 11).sum()} de {len(prov_n)} "
      f"({(prov_n < 11).mean():.1%})")
print(t_z.round(2))

atipico_alto = r["nivel_monto"] == "muy alto"; dentro = r["log_monto"].between(cerca_inf, cerca_sup)
tabla_z = pd.crosstab(np.where(atipico_alto, "por encima de la cerca", "dentro de las cercas")[atipico_alto | dentro],
                      r.loc[atipico_alto | dentro, "con_friccion"])
chi2_z, p_z, _, esp_z = stats.chi2_contingency(tabla_z)
print(f"\nPor encima de la cerca superior: {r.loc[atipico_alto, 'con_friccion'].mean():.2%} vs dentro de las cercas: "
      f"{r.loc[dentro, 'con_friccion'].mean():.2%} | chi2 = {chi2_z:.1f}, {fmt_p(p_z)}, V de Cramér = {cramers_v(tabla_z):.3f}, "
      f"frecuencia esperada mínima = {esp_z.min():.0f}")
dif, lo, hi, t, gl, p = welch(r.loc[r["con_friccion"] == 1, "desvio_log"].abs(), r.loc[r["con_friccion"] == 0, "desvio_log"].abs())
print(f"Distancia absoluta a la mediana del proveedor (log10), con vs sin fricción: diferencia de medias = {dif:.3f}, "
      f"IC 95% [{lo:.3f}, {hi:.3f}], {fmt_p(p)}")

fig, ax = plt.subplots(figsize=(10, 4))
ax.errorbar(orden_n, t_z["tasa_pct"], yerr=[t_z["tasa_pct"] - t_z["ic95_inf"], t_z["ic95_sup"] - t_z["tasa_pct"]],
            fmt="o", color="steelblue", capsize=5)
ax.axhline(r["con_friccion"].mean() * 100, color="gray", ls="--", lw=1, label="tasa de estas facturas")
ax.set_title("Tasa de fricción según qué tan atípico es el monto para su proveedor (cercas del IQR), con IC 95%")
ax.set_xlabel("Nivel del monto frente a la historia del proveedor"); ax.set_ylabel("% de facturas con fricción"); ax.legend()
plt.tight_layout(); plt.show()
```

La hipótesis 4 se apoya. Las facturas que quedan por encima de la cerca superior de su proveedor tienen una tasa de fricción de 9,69%, frente a 4,96% de las que están dentro de las cercas (chi² = 64,9; p < 0,001). Es casi el doble, aunque la V de Cramér es pequeña (0,037) porque estas facturas son pocas. La prueba t de Welch muestra lo mismo con una medida continua, ya que las facturas con fricción están, en promedio, 0,068 unidades de log10 más lejos de la mediana de su proveedor que las aprobadas (IC 95% de 0,053 a 0,083; p < 0,001).

Algo que no esperabamos es que las facturas del nivel muy bajo también tienen más fricción (6,16%). Es decir, la tabla tiene forma de U y lo atipico en cualquier dirección se asocia con más problemas. Por eso, la distancia del monto a lo habitual del proveedor parece una buena variable para el modelo, siempre que entre en valor absoluto o con las dos colas.

Monto vs canal (cuantitativa vs categórica)

Como complemento, comparamos la distribucion del monto aprobado entre canales con un boxplot y con un ANOVA de una vía sobre el log10 del monto, dado que en esa escala la distribución es mucho más simétrica.

Esto lo controlamos en el modelo multivariado.

### 5.6 Análisis multivariado

a) Correlaciones entre variables cuantitativas

Usamos solo variables que se conocen al momento de crear la factura, es decir, sin los campos de aprobacion por la fuga de informacion. Además, excluimos los extremos no verificados.

```python
# los extremos no verificados quedan por fuera de r
r["antiguedad_proveedor_dias"] = (r["Invoice Created Date"] - r["proveedor_id"].map(primera)).dt.days
vars_corr = ["log_monto", "desvio_log", "n_prov", "antiguedad_proveedor_dias", "con_friccion"]
corr = r[vars_corr].corr(method="spearman")
print(corr.round(2))
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="RdBu_r", vmin=-1, vmax=1)
plt.title("Correlación de Spearman entre variables cuantitativas (facturas resueltas)")
plt.tight_layout(); plt.show()

con_ext = resueltas.copy()
con_ext["antiguedad_proveedor_dias"] = (con_ext["Invoice Created Date"] - con_ext["proveedor_id"].map(primera)).dt.days
con_ext = con_ext.dropna(subset=["antiguedad_proveedor_dias"])
sin_ext = con_ext[~con_ext["extremo_no_verificado"]]
for nombre, d in [("con extremos no verificados", con_ext), ("sin extremos no verificados", sin_ext)]:
    print(f"monto vs antigüedad del proveedor ({nombre}): Pearson = "
          f"{stats.pearsonr(d['monto_cad'], d['antiguedad_proveedor_dias'])[0]:.4f} | Spearman = "
          f"{stats.spearmanr(d['monto_cad'], d['antiguedad_proveedor_dias'])[0]:.4f} | media del monto = {d['monto_cad'].mean():,.0f} CAD")
```

Ninguna variable cuantitativa por sí sola tiene una asociación importante con la fricción, ya que todas las correlaciones con ella están entre menos 0,04 y 0,04. Entre las variables, las unicas relaciones relevantes son la del monto con su distancia a la mediana del proveedor (0,70), que es esperable porque esa distancia se construye con el monto, y la del número de facturas del proveedor con su antigüedad (0,30).

La sensibilidad a los etxremos no verificados muestra algo importante. La correlación de Pearson entre el monto y la antigüedad del proveedor pasa de 0,0450 a 0,0521 al quitarlos, y la media del monto baja de157.579 a 149.852 CAD. En cambio, la de Spearman casi no cambia (0,1190 freente a 0,1188) porque trabaja con rangos. Esto confirma que para estas variables Spearman es la medida adecuada, y que los extremos sí afectan los estadísticos que dependen de la magnitud.

b) Vista conjunta de canal, categoria y gasto

El mapa de calor cruza tres variables: el canal, la categoría y el porcentaje que representa cada categoría dentro del gasto aprobado del canal. Además, lo complementamos con una prueba chi cuadrado sobre el número de faccturas.

```python
top_cat = aprobadas.groupby("Commodity Level 1")["monto_cad"].sum().nlargest(7).index
cat_g = aprobadas["Commodity Level 1"].where(aprobadas["Commodity Level 1"].isin(top_cat), "Otras categorías")
canales_p = ["csp", "sftp", "ui", "email", "cxml"]
sub_h = aprobadas["Invoice Channel"].isin(canales_p)
mix = pd.crosstab(cat_g[sub_h], aprobadas.loc[sub_h, "Invoice Channel"], values=aprobadas.loc[sub_h, "monto_cad"],
                  aggfunc="sum", normalize="columns")[canales_p] * 100
plt.figure(figsize=(10, 5.5))
sns.heatmap(mix, annot=True, fmt=".0f", cmap="Blues", cbar_kws={"label": "% del gasto aprobado del canal"})
plt.title("Composición del gasto aprobado de cada canal por categoría (%)")
plt.xlabel("Canal de ingreso"); plt.ylabel("Categoría"); plt.yticks(rotation=0)
plt.tight_layout(); plt.show()
tabla_cc = pd.crosstab(cat_g[sub_h], aprobadas.loc[sub_h, "Invoice Channel"])
chi2_cc, p_cc, _, esp_cc = stats.chi2_contingency(tabla_cc)
print(f"Categoría vs canal (conteos): chi2 = {chi2_cc:,.1f}, {fmt_p(p_cc)}, V de Cramér = {cramers_v(tabla_cc):.3f}, "
      f"frecuencia esperada mínima = {esp_cc.min():.0f}")
```

Cada canal tiene un perfil de compra distinto, y la asociación entre categoría y canal es fuerte (V de Cramér de 0,452). La frecuencia esperada mínima es 36, así que se cumple el supuesto de la prueba. En email, el 67% del gasto no tiene categoría y en ui es el 54%, frente al 30% del portal y de sftp. Por lo tanto, la falta de categoría se concentra en las facturas que digita el Banco. Por otro lado, en cxml el 73% del gasto es Advertising & Marketing, es decir que ese canal es casi exclusivo de los proveedores de publicidad, y Human Resources / Personnel pasa sobre todo por el portal (14% de su gasto).

Esto muestra que el canal y la categoria no son independientes. Por eso, en el modelo conviene usarlas juntas o con interacciones, y cualquier comparación entre categorías se debería hacer dentro de cada canal.

c) Regresión lineal de la friccion

La regresión lineal permite ver la asociación de cada variable con la fricción manteniendo constantes las demás. La variable respuesta vale 100 si la factura tuvo fricción y 0 si no, así que cada coeficiente se lee directamente en puntos porcentuales. Incluimos el canal (con csp como referencia), el log10 del monto, si tiene contrato, si el proveedor es intragrupo, si la moneda es CAD, la antigüedad del proveedor y el semestre, por la deriva que vimos en la sección 2. Los coeficientes los leemos como asociaciones condicionales y no como efetos causales.

Hay que tener en cuenta dos cosas. Primero, la regresión supone que las facturas son independientes y no lo son, porque las de un mismo proveedor se parecen. En la sección 5.2 el bootstrap por proveedor dio un intervalo unas 1,7 veces más ancho que el queasume independencia. Así que, los intervalos de esta regresión son optimistas y los resultados con p cercano a 0,05 se deben leer con cuidado. Segundo, la respuesta es binaria, por lo tanto el R² va a ser bajo y la regresion sirve para comparar asociaciones y no para predecir.

```python
# respuesta 100/0 para leer los coeficientes en puntos porcentuales
m = resueltas[~resueltas["extremo_no_verificado"]].dropna(subset=["proveedor_id"]).copy()
m["friccion_pct"] = m["con_friccion"] * 100
m["log_monto"] = np.log10(m["monto_cad"])
m["semestre"] = m["Invoice Created Date"].dt.year.astype(str) + " S" + np.where(m["Invoice Created Date"].dt.month <= 6, "1", "2")
m["canal"] = m["Invoice Channel"].where(m["Invoice Channel"].isin(["csp", "sftp", "ui", "email", "cxml"]), "otro")
m["intragrupo"] = m["prov_intragrupo"].astype(int)
m["contrato"] = m["tiene_contrato"].astype(int)
m["moneda_cad"] = (m["Invoice Currency Code"] == "CAD").astype(int)
m["antiguedad_anios"] = (m["Invoice Created Date"] - m["proveedor_id"].map(primera)).dt.days / 365.25
modelo = smf.ols("friccion_pct ~ C(canal, Treatment('csp')) + log_monto + contrato + intragrupo + moneda_cad "
                 "+ antiguedad_anios + C(semestre)", data=m).fit()
coef = pd.DataFrame({"coef_puntos_pct": modelo.params, "ic95_inf": modelo.conf_int(ALPHA)[0],
                     "ic95_sup": modelo.conf_int(ALPHA)[1], "p_valor": modelo.pvalues})
coef.index = (coef.index.str.replace(r"C\(canal, Treatment\('csp'\)\)\[T\.", "canal=", regex=True)
              .str.replace(r"C\(semestre\)\[T\.", "semestre=", regex=True).str.replace("]", ""))
print(f"Facturas: {int(modelo.nobs):,} | R² = {modelo.rsquared:.3f} | R² ajustado = {modelo.rsquared_adj:.3f}")
coef.round(3)
```

El R² es 0,016, lo que quiere decir quelas variables disponibles al crear la factura explican muy poco de la fricción. Este resultado es importante porque muestra que el riesgo no se puede anticipar solo con estas características generales. Es decir, hacen falta las variables específicas de duplicado de la sección 3.5 y etiquetas reales.

Manteniendo las demás variables constantes, la asociación más fuerte es la de la moneda. Las facturas en CAD tienen 5,91 puntos porcentuales menos de friccion que las de otras monedas (IC 95% de 5,38 a 6,45 puntos), lo cuál es una diferencia grande si se compara con la tasa global de 5,33%. Esto sugiere que la conversión, los proveedores del exterior o las entidades fuera de Canadá generan más correcciones. Además, cada vez que el monto se multiplica por 10, la fricción sube 1,42 puntos (IC de 1,04 a 1,80).

En cuanto al canal, email (4,87 puntos menos) y ui (2,31 puntos mneos) tienen menos fricción que el portal, y sftp también queda por debajo del portal (1,30 puntos menos, IC de 0,76 a 1,83). Esto significa que la tasa alta de sftp en el análisis bivariado se explicaba por su mezcla de monedas y montos. Por otro lado, ser intragrupo se asocia con 1,83 puntos menos de fricción, al revés de lo que veíamos en la sección 5.4, lo que muestra que esa diferencia tambien venía de otras variables. Tener contrato (0,54 puntos menos; p = 0,04) y la antigüedad del proveedor (0,25 puntos más por año; p = 0,03) quedan cerca del límite, así que con los intervalos optimistas no se deben tomar como asociaciones firmes. Por último, el segundo semestre de 2025 tiene 1,90 puntos menos de fricción que el último de 2024, lo cuál confirma que era necesario controlar el periodo.

d) Reducción de dimensionalidad (PCA) de los perfiles de proveedor

Describimos cada proveedor con al menos 10 facturas resueltas con ocho variables: volumen, monto tipico, dispersión del monto, porcentaje sin contrato y su mezcla de canales y monedas. Estandarizamos las variables y las proyectamos en dos componentes. Después, coloreamos los puntos segun la tasa de fricción para ver si el riesgo se agrupa en algún tipo de proveedor. Los campos de aprobación no los incluimos.

```python
perfil = resueltas.groupby("proveedor_id").agg(
    log_facturas=("monto_cad", lambda s: np.log10(len(s))),
    log_mediana_monto=("monto_cad", lambda s: np.log10(s.median())),
    dispersion_log_monto=("monto_cad", lambda s: np.log10(s).std(ddof=0)),
    pct_sin_contrato=("tiene_contrato", lambda s: 1 - s.mean()),
    pct_sftp=("Invoice Channel", lambda s: (s == "sftp").mean()),
    pct_csp=("Invoice Channel", lambda s: (s == "csp").mean()),
    pct_ui=("Invoice Channel", lambda s: (s == "ui").mean()),
    pct_moneda_cad=("Invoice Currency Code", lambda s: (s == "CAD").mean()),
    tasa_friccion=("con_friccion", "mean"),
    facturas=("con_friccion", "size"))
perfil = perfil[perfil["facturas"] >= 10]
X_cols = [c for c in perfil.columns if c not in ["tasa_friccion", "facturas"]]
pca = PCA(n_components=2, random_state=RANDOM_STATE)
comp = pca.fit_transform(StandardScaler().fit_transform(perfil[X_cols]))
print("Varianza explicada:", pca.explained_variance_ratio_.round(3), "| total:", pca.explained_variance_ratio_.sum().round(3))
cargas = pd.DataFrame(pca.components_.T, index=X_cols, columns=["PC1", "PC2"])
print(cargas.round(2))
fig, ax = plt.subplots(figsize=(10, 6))
sc = ax.scatter(comp[:, 0], comp[:, 1], c=perfil["tasa_friccion"] * 100, cmap="viridis", s=14, vmax=30)
plt.colorbar(sc, label="% de facturas con fricción (tope visual 30%)")
ax.set_title("Proveedores en las dos primeras componentes principales, coloreados por tasa de fricción")
ax.set_xlabel(f"PC1 ({pca.explained_variance_ratio_[0]:.0%} de la varianza)"); ax.set_ylabel(f"PC2 ({pca.explained_variance_ratio_[1]:.0%})")
plt.tight_layout(); plt.show()
print("Spearman tasa de fricción vs PC1:", round(stats.spearmanr(comp[:, 0], perfil["tasa_friccion"])[0], 3),
      "| vs PC2:", round(stats.spearmanr(comp[:, 1], perfil["tasa_friccion"])[0], 3))
```

Las dos primeras componentes explican el 50,3% de la varianza (27,8% y 22,5%). La PC1 separa a los proveedores del portal con contrato y montos variables (cargas positivas en el porcentaje de csp y en la dispersion del monto) de los proveedores sin contrato y digitados internamente (cargas negativas en el porcentaje sin contrato y en el porcentaje de ui). En cambio, la PC2 está dominada por el uso de sftp.

La tasa de fricción del proveedor se correlaciona más con la PC1 (ρ = 0,395) que con la PC2 (ρ = 0,228). Además, en la gráfica vemos que los proveedores con más fricción (los puntos claros) se ubican hacia laderecha, en el perfil del portal con contrato. Esto coincide con lo que vimos en la sección 3.5, ya que buena parte de la fricción son correcciones y reenvíos de proveedores que facturan por el portal. Es una friccion operativa que no se debe confundir con el riesgo de un duplicado pagado.

### 5.7 Suficiencia de los datos para el modelo

El enunciado pide validar la calidad y también la suficiencia de los datos para la solución planteada.
```python
hist_prov = aprobadas.groupby("proveedor_id").size()
prov5 = hist_prov[hist_prov >= 5].index
cubre = aprobadas[aprobadas["proveedor_id"].isin(prov5)]
print(f"Proveedores con >= 5 facturas aprobadas en la ventana: {len(prov5):,} de {hist_prov.size:,} ({len(prov5) / hist_prov.size:.1%}); "
      f"cubren {len(cubre) / len(aprobadas):.1%} de las facturas y {cubre['monto_cad'].sum() / aprobadas['monto_cad'].sum():.1%} del gasto")

# peor caso p = 0,5 para el tamaño de muestra
z = stats.norm.ppf(0.975)
for margen in [0.05, 0.07, 0.10]:
    print(f"Margen de error ±{margen:.0%}: {int(np.ceil(z**2 * 0.25 / margen**2))} casos etiquetados")

estrato = np.select(
    [ap_v["dup_numero"],
     ap_v["dias_desde_mismo_monto"].eq(0) & ap_v["mismo_monto_mismo_transit"],
     ap_v["dias_desde_mismo_monto"].eq(0),
     ap_v["dias_desde_mismo_monto"].between(1, 7)],
    ["A. Número casi idéntico", "B. Mismo monto, mismo día y mismo transit", "C. Mismo monto, mismo día, otro transit",
     "D. Mismo monto en 1 a 7 días"], default="Sin señal")
estratos = pd.Series(estrato, index=ap_v.index).value_counts().drop("Sin señal").sort_index().to_frame("facturas")
estratos["muestra_propuesta"] = [min(n, k) for n, k in zip(estratos["facturas"], [100, 70, 30, 70])]
print("Tamaño total de la muestra propuesta:", estratos["muestra_propuesta"].sum())
estratos
```

Solo 961 de los 2.775 proveedores (el 34,6%) tienen 5 o más facturas aprobadas en la ventana. Sin embargo, esos proveedores cubren el 93,5% de las facturas y el 93,1% del gasto. Es decir, las comparaciones contra la historia del proveedor, como la distancia a lo habitual del proveedor o la periodicidad, sí se pueden hacer para casi todo el gasto. En cambio, para los proveedores con poca historia hay que usar reglas más simples, como la similitud del número y elmismo monto.

En cuanto al etiquetado, para estimar la precisión de las alertas con un 95% de confianza y un margen de 7 puntos porcentuales necesitamos 196 casos en el peor escenario (p = 0,5). Con un margen de 5 puntos harian falta 385, y con uno de 10 puntos solo 97. Con esto armamos una muestra estratificada de 202 casos. Esta incluye las 32 facturas con numero casi idéntico, que son las más sospechosas, 70 del estrato de mismo monto, mismo día y mismo transit, 30 del estrato demismo día en otro transit y 70 del estrato de mismo monto entre 1 y 7 días. Estratificar asegura que haya casos de cada patrón y permite estimar la precisión de cada regla por separado.

En conclusión, los datos son suficientes en volumen y cobertura para construir las variables. Lo que falta es la etiqueta, y ese es el insumo que el Banco tiene que aportar antes de la entrega 2.
